```python
#digit concept
a=8919
#use %10 to get last digit
a%10
```

```python
#use //10 or int(/10) to remove last digit
a//10
```

```python
int(a/10)
```

```python
number=a
while(number>0):
    lastdigit=number%10
    number=number//10
    print(lastdigit)
```

```python
#counting no.of digits in a number
number=a
count=0
while(number>0):
    count+=1
    number//=10
count
    
```

```python
import math
count=int(math.log(a,10))+1
count
```
