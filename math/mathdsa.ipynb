```python
#armstrong number
import math
number=int(input("Enter number : "))
digits=int(math.log(number,10))+1
digits
def armst_num(number):
    num=0
    number2=number
    while(number2>0):
        num+=(number2%10)**digits
        number2//=10
    if(num==number):
        print(f"{number} is an armstrong number")
    else:
        print(f"{number} is not a armstrong number")
armst_num(number)
```

```python
#all divisiors more optimal using sqrt
import math
number=int(input("Enter Number : "))
def all_divisiors2(number):
    divisiors=[]
    for i in range(1,int(math.sqrt(number))+1):
        if(number%i==0):
            divisiors.append(i)
            if(number/i!=i):
                divisiors.append(int(number/i))
    return sorted(divisiors)
div=all_divisiors2(number)
    
```

```python
def prime_num(number):
    div=all_divisiors2(number)
    if(len(div)==2 ):
        print(f"{number} is a prime number")
    else:
        print(f"{number} is not a prime number")
prime_num(7)
```
